```python
from Bio import SeqIO
import random
import os
os.chdir("D:/jupyter code/MPOGAN-main/MPOGAN-main/logs/MPOGAN_finetuning")
os.getcwd()
```

```python
# 输入文件和输出文件
input_fasta = "epoch700_outputs_50000.fasta"
output_fasta = "sampled_2.fasta"

# 设定长度区间
min_len = 16
max_len = 25

# 读取所有序列并过滤
filtered_seqs = [
    record for record in SeqIO.parse(input_fasta, "fasta")
    if min_len <= len(record.seq) <= max_len
]

print(f"符合长度要求的序列数: {len(filtered_seqs)}")

# 检查是否足够抽样
sample_size = 2000
if len(filtered_seqs) < sample_size:
    print(f"警告: 仅有 {len(filtered_seqs)} 条符合条件的序列，将全部输出。")
    sampled_seqs = filtered_seqs
else:
    sampled_seqs = random.sample(filtered_seqs, sample_size)

# 写入抽样结果
SeqIO.write(sampled_seqs, output_fasta, "fasta")
print(f"已将 {len(sampled_seqs)} 条序列写入 {output_fasta}")
```

```python
import os

def split_peptides(input_file):
    """简洁版本 - 基于输入文件名生成输出文件"""
    groups = {"_5-15": (5,15), "_16-25": (16,25), 
              "_26-35": (26,35), "_36-50": (36,50)}
    
    base = os.path.splitext(input_file)[0]
    data = {g: [] for g in groups}
    
    with open(input_file) as f:
        h, s = "", ""
        for l in f:
            if l.startswith(">"):
                if h: 
                    le = len(s)
                    for g, (mi, ma) in groups.items():
                        if mi <= le <= ma: data[g].append((h, s))
                h, s = l.strip(), ""
            else: s += l.strip()
        if h: 
            le = len(s)
            for g, (mi, ma) in groups.items():
                if mi <= le <= ma: data[g].append((h, s))
    
    for g, seqs in data.items():
        out_file = f"{base}_{g}.fasta"
        with open(out_file, "w") as f:
            for h, s in seqs: f.write(f"{h}\n{s}\n")
        print(f"{out_file}: {len(seqs)} sequences")

# 使用
split_peptides("./data/output/fasta/mmcd_moco_esmc_pretrain_dynamic_length_peptides.fasta")
```
